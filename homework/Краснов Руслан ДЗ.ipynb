```python
import numpy as np

class Minimizer:
    def __init__(self, f, df, alpha=0.01, n_iter=1000):
        self.f = f
        self.df = df
        self.alpha = alpha
        self.n_iter = n_iter
        self.trace = []

    def move(self, x):
        x = np.array(x)
        grad_vec = np.array(self.df(x))
        self.trace.append([x, grad_vec])
        return x - self.alpha * grad_vec

    def run(self, start_x):
        x = np.array(start_x)

        lowest_val = self.f(x)
        optimal_x = np.copy(x)
        
        for k in range(self.n_iter):
            curr_val = self.f(x)

            self.trace.append({
                "iteration": k,
                "params": x,
                "value": curr_val,

            })

            if curr_val < lowest_val: 
                min_vaue = curr_val
                optimal_x = np.copy(x)

            grad_vec = self.df(x)
            
            if np.linalg.norm(grad_vec) < 1e-6:
                print(f"схождение при значении переменных {x}, значение функции {grad_vec}")
                break

            x = self.move(x)

        end_val = self.f(x)
        self.trace.append({
            "iteration": self.n_iter,
            "params": np.copy(x),
            "value": end_val
        })

        if end_val < lowest_val:
            lowest_val = end_val
            best_paramns = np.copy(x)

        return lowest_val, optimal_x

    def show_trace(self, interval=1):
        for i in self.history:
            print(f"итерация: {i["iteration"]}, значение переменнхы {i["params"]}, значение {i["value"]}")
```

```python
def cost_func(x):
    return x[0]**2 + x[1]**2

def df(x):
    return 2*x[0] + 2*x[1]

optimizer = Minimizer(cost_func, df)

start = [50.0, -40.0]
optimizer.run(start)
```
